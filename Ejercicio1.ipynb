```python
import math
from scipy.optimize import fsolve
```

```python
prob_bin = 0.6
prob_no_bin = 0.4

def prob_0_o_1(N):
    p0 = prob_bin ** N
    p1 = N * prob_no_bin * (prob_bin ** (N - 1))
    return p0 + p1
```

```python
def ecuacion(N):
    return prob_0_o_1(N) - 0.01

estimacion = math.log(0.01) / math.log(prob_bin)
n_solucion = fsolve(ecuacion, estimacion)[0]
n_minimo = math.ceil(n_solucion)

print(f"Solución numérica: n ≈ {n_solucion:.2f}")
print(f"Mínimo entero n: {n_minimo}")
```
